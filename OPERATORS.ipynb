```python
N = 40
M = 80
print(M == N )
print(M != N )
print(M > N)
print(M < N)
print(M >= N)
print(M <= N)



```

```python
N = 40 + 12j
M = 80 + 33j
print(M == N )
print(M != N )
print(abs(M) > abs(N))
print(abs(M) < abs(N))
print(abs(M) >= abs(N))
print(abs(M) <= abs(N))
# abs means absolute value
#abs()is a built in function that return the absolute value of a number . for a complex number, it return its magnitude.

```
